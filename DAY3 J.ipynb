```python
%%writefile calculator_module.py

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b
```

```python
import calculator_module

print("Addition:", calculator_module.add(10, 5))

print("Subtraction:", calculator_module.subtract(10, 5))

print("Multiplication:", calculator_module.multiply(10, 5))

print("Division:", calculator_module.divide(10, 5))
```
